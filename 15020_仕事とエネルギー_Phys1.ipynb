### エネルギー

- エネルギーとは(物理的な)仕事をする能力.
- 仕事という物理量は, 単位 J で表される.
  - 同じ単位の物理量は, **方向の情報は忘れて、** 加法減法が容易にできる.
  - エネルギーの単位も,  単位 J となることが, 後でわかります
- 物体に蓄えられるエネルギーは, 仕事と同様に
  > 力 × 動かした距離

  で与えられる.

- 物体が持つエネルギーは,
  - 物体が外部から仕事を加えられれば増加し,
  - 物体が外部に対して仕事をすれば減少する
